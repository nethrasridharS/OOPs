```python
#OOPS_ASSIGNMENT
class Car:
    def __init__ (self,make,model,color,mileage,number_plate):
        self.make = make
        self.model = model
        self.color = color
        self.mileage = mileage
        self.number_plate = number_plate

    def greeting(self):
        print (f"Thanks for purchasing {self.model} model car. Made by {self.make} brand with a color of {self.color} that gives {self.mileage} mileage")
        print(f"and {self.number_plate} is your fancy number")

SoldCar1 =Car("Hyundai","Creta","Lavender","18kmpl","TN 01 AB 0001")
SoldCar1.greeting()

SoldCar2 = Car("Tata","Nexon","Black","17kmpl" , "AP 02 AA 00000")
SoldCar2.greeting()

SoldCar3 = Car("Maruti Suzuki","Swift","Navy BLue","20kmpl","DL 05 KB 0202")
SoldCar3.greeting()
```
